```python
import numpy as np
import matplotlib.pyplot as plt
import glob
import os

# Folder containing your datasets
folder = "./training_data"

# Find all CSV and TXT files
files = sorted(
    glob.glob(os.path.join(folder, "*.csv"))
)

print(f"Found {len(files)} files")

data = np.loadtxt(files[433], delimiter=',')

fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=data[:,0],
    y=data[:,1],
    z=data[:,2],
    mode='lines+markers',
    line=dict(color='royalblue', width=5),
    marker=dict(size=3, color='red'),
    name='3D Number'
))

fig.update_layout(
    title='3D Coordinate Plot',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='data'
    )
)

fig.show()


# Use your existing data array
fig, ax = plt.subplots(figsize=(9, 9))

ax.plot(
    data[:, 0],
    data[:, 1],
    '-o',
    color='royalblue',
    markersize=4,
    linewidth=2
)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('2D Projection of the Number (XY Plane)')

ax.set_aspect('equal', adjustable='datalim')
ax.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
